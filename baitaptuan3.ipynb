Bài 1: Nhập vào một số nguyên dương có 3 chữ số. Tìm và in ra màn hình “SỐ 
NGUYÊN” đảo ngược của n. Không dùng kiểu chuỗi. Chú ý: kiểm tra chỉ xử 
lý các số nguyên dương có đúng 3 chữ số. Các số khác không xử lý.

```python
n = int(input())
if 100 <= n <= 999:
    a = n // 100
    b = (n // 10) % 10
    c = (n % 100) % 10
    print(c * 100 + b * 10 + a)
    print("Error")
```

Bài 11: [Phân loại tam giác] 
Nhập vào 3 số nguyên dương a,b,c là độ dài 3 cạnh của một tam giác. Kiểm 
tra 3 số đã nhập có tạo thành một tam giác hợp lệ hay không. Nếu không in 
ra "Không phải tam giác". Nếu có, phân loại tam giác đó là "Tam giác đều", 
"Tam giác cân", hay "Tam giác thường".

```python
a,b,c=map(int,input().split())
if a > 0 and b > 0 and c > 0 and a == b == c:
    print("Tam giác đều") 
elif a == b or b == c or c == a:
    print("Tam giác cân")
elif a + b > c and a + c > b and b + c > a:
    print("Tam giác thường")
else:
    print("Không phải tam giác")
```

Bài 14: [Giải phương trình bậc nhất] 
Nhập vào hai số thực a và b của phương trình ax + b = 0. Tìm và in ra 
nghiệm của phương trình, kết quả làm tròn đến số thập phân thứ 2. In ra "Vô 
nghiệm" hoặc "Vô số nghiệm" trong trường hợp tương ứng.

```python
a = float(input())
b = float(input())
if a == b == 0:
    print("Vô số nghiệm")
elif a == 0 and b != 0:
    print("Vô nghiệm")
else:
    print(round(-b/a,2))
```

Bài 15: [Xếp loại học lực] 
Nhập vào điểm trung bình của một học sinh và in ra học lực của học sinh đó. 
Xếp loại học lực theo quy tắc: - >= 8.0: Giỏi - >= 6.5: Khá - >= 5.0: Trung bình - < 5.0: Yếu 

```python
n = float(input())
if n >= 0 and n <= 10:
    if n >= 8.0:
        print("Giỏi")
    elif n >= 6.5:
        print("Khá")
    elif n >= 5.0:
        print("Trung binh") 
    else: 
        print("Yếu")
else:
    print("Error")
```

Bài 16: [Rounding] 
Nhập vào một số thực: in ra kết quả làm tròn lên, tròn xuống, làm tròn tới số 
nguyên gần nhất (không được sử dụng hàm sẵn có của pyton). Các kết quả 
cách nhau một dấu cách

```python
n = float(input())
a = n // 1 + 1
b = n // 1
if n - b >= 0.5:
    n = a
else:
    n = b
print(a,b,n, sep=" ")
```
