Bài 3: Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là 
độ dài ba cạnh hay không, nếu  không thì in  ra màn  hình  'Khong la ba  canh 
cua mot tam  giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, 
Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn 
hình.

```python
a,b,c=map(int,input().split())
p = (a+b+c)/2
s = (p*(p-a)*(p-b)*(p-c))**1/2
deu = a == b == c
can = a == b or b == c or c == a
vuong = a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2
if a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a:
    if deu:
        x = "Tam giac deu"
    elif can and vuong:
        x = "Tam giac vuong can"
    elif can:
        x = "Tam giac can"
    elif vuong:
        x = "Tam giac vuong"
    else:
        x = "Tam giac thuong"
    print(x, s, 2*p, sep=",")
else:
    print("Khong la ba canh cua mot tam giac")
```

Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0) 

```python
a = float(input())
b = float(input())
c = float(input())
d = b ** 2 - (4 * a * c)
if a == b == c == 0:
    print("Vo so nghiem")
elif d > 0:
    x1 = (-b + (d ** 1/2)) / 2 * a
    x2 = (-b - (d ** 1/2)) / 2 * a
    print(x1, x2)
elif d == 0:
    x = -b / 2 * a
    print(x)
else:
    print("Vo nghiem")
```
